## 🟢 Section A — Warm-up

### Exercise A1 — Your first function

```python
# A1 - define it once, then call it as often as you like
def greet():
    """Print a fixed greeting."""
    print("Hello from a function!")

greet()
greet()
```

### Exercise A2 — A function with a parameter

```python
# A2 - 'name' is filled in by whatever the caller passes
def greet_person(name):
    """Print a greeting for one person."""
    print(f"Hello, {name}!")

greet_person("Ana")
greet_person("Bruno")
```

### Exercise A3 — `return` instead of `print`

```python
# A3 - returning the value means the caller can use it, not just look at it
def double(number):
    """Return twice the number given."""
    return number * 2

print(double(7))
print(double(7) + double(3))    # only possible because double() RETURNS
```

### Exercise A4 — Two parameters

```python
# A4 - parameters are filled in by position, left to right
def rectangle_area(width, height):
    """Return the area of a rectangle."""
    return width * height

print(rectangle_area(4, 5))
print(rectangle_area(2.5, 3))
```

### Exercise B1 — A default argument

```python
# B1 - the default is used whenever the caller leaves it out
def add_vat(amount, rate=23):
    """Return the amount with VAT added, rounded to 2 decimals."""
    return round(amount + (amount * rate / 100), 2)

print(add_vat(100))
print(add_vat(100, 6))
print(add_vat(59.90))
```

### Exercise B2 — Keyword arguments

```python
# B2 - naming the values means their order no longer matters
def book_room(nights, guests=1, breakfast=False):
    """Return a one-line description of a booking."""
    return f"nights: {nights}, guests: {guests}, breakfast: {breakfast}"

print(book_room(3))
print(book_room(3, 2))
print(book_room(nights=3, breakfast=True))
print(book_room(breakfast=True, guests=4, nights=2))
```

### Exercise B3 — Local and global

```python
# B3 - 'saving' is local: it exists only while the function runs.
# 'discount_rate' is global, so the function can read it.
discount_rate = 10

def apply_discount(price):
    """Take the standard discount off a price."""
    saving = price * discount_rate / 100
    return round(price - saving, 2)

print(apply_discount(80))
print(discount_rate)            # unchanged by the call

# Asking for 'saving' out here would be an error, because it no longer exists:
# print(saving)
print("saving does not exist outside the function")
```

### Exercise B4 — A function that answers yes or no

```python
# B4 - the comparison already produces True or False, so return it directly
def is_expensive(price, limit=50):
    """Return True when the price is above the limit."""
    return price > limit

print(is_expensive(80))
print(is_expensive(20))
print(is_expensive(20, 10))

# Because it returns a bool, it can be used straight inside an if
if is_expensive(120):
    print("Needs approval")
```

### Exercise C1 — A function that takes a list

```python
# C1 - the running-total pattern from Level 1, now with a name
def total_of(numbers):
    """Return the total of a list of numbers."""
    running = 0
    for number in numbers:
        running += number
    return round(running, 2)

print(total_of([8.50, 12.00, 2.10]))
print(total_of([100, 200]))
print(total_of([]))             # an empty list totals 0
```

### Exercise C2 — A function that takes a dictionary

```python
# C2 - reaching into the dict by key, inside the function
def describe_customer(customer):
    """Return a one-line description of a customer."""
    name = customer["name"]
    city = customer["city"]
    orders = customer["orders"]
    return f"{name} from {city} has {orders} orders"

ana = {"name": "Ana Silva", "city": "Porto", "orders": 3}
bruno = {"name": "Bruno Costa", "city": "Lisbon", "orders": 12}

print(describe_customer(ana))
print(describe_customer(bruno))
```

### Exercise C3 — A function that returns a dictionary

```python
# C3 - a function can hand back a whole dict, not just one value
def describe_numbers(numbers):
    """Return a dict of basic statistics about a list of numbers."""
    return {
        "count": len(numbers),
        "total": round(sum(numbers), 2),
        "smallest": min(numbers),
        "largest": max(numbers),
        "average": round(sum(numbers) / len(numbers), 2),
    }

stats = describe_numbers([8.50, 12.00, 2.10, 45.00])

print(stats)
print(stats["average"])         # the result is a normal dict
```

### Exercise D1 — Functions calling functions

```python
# D1 - three small jobs, and one function that puts them in order
def clean_title(title):
    """Remove stray spaces from a book title."""
    return title.strip()

def shipping_cost(weight_kg):
    """Return the shipping cost: 2.50 EUR per kg, minimum 5.00."""
    cost = weight_kg * 2.50
    if cost < 5.00:
        return 5.00
    return round(cost, 2)

def describe_parcel(parcel):
    """Build one line describing a parcel, using the two helpers above."""
    title = clean_title(parcel["title"])
    cost = shipping_cost(parcel["weight_kg"])
    return f"{title:<20}{cost:>7.2f} EUR"

parcels = [
    {"title": "  Dune ", "weight_kg": 0.8},
    {"title": "Python Basics", "weight_kg": 3.0},
    {"title": " Atlas of Ports", "weight_kg": 12.4},
]

for parcel in parcels:
    print(describe_parcel(parcel))
```

### Exercise D2 — A complete library report

```python
# D2 - the rules live in named variables at the top, so one edit changes
# the whole report
SHORT_LIMIT = 200      # up to here it is a short book
LONG_LIMIT = 500       # above here it is long


def clean_title(title):
    """Remove stray spaces from a title."""
    return title.strip()


def classify_length(pages):
    """Label a book short, medium or long."""
    if pages > LONG_LIMIT:
        return "long"
    elif pages > SHORT_LIMIT:
        return "medium"
    else:
        return "short"


def describe_book(book):
    """Build one line of the report for a single book."""
    title = clean_title(book["title"])
    return f"{title:<18}{book['pages']:>5}  {classify_length(book['pages'])}"


def summarise_books(books):
    """Count the books by length and total their pages. Returns a dict."""
    counts = {"short": 0, "medium": 0, "long": 0}
    pages = 0
    for book in books:
        counts[classify_length(book["pages"])] += 1
        pages += book["pages"]
    counts["total_pages"] = pages
    return counts


books = [
    {"title": "  Dune ", "pages": 412},
    {"title": "Python Basics", "pages": 180},
    {"title": " Atlas of Ports", "pages": 640},
    {"title": "Short Stories ", "pages": 95},
    {"title": "The Long Road", "pages": 720},
]

print("LIBRARY REPORT")
print("-" * 34)
for book in books:
    print(describe_book(book))

print("-" * 34)
summary = summarise_books(books)
for size in ["short", "medium", "long"]:
    print(f"{size:<14}{summary[size]:>5}")
print(f"{'total pages':<14}{summary['total_pages']:>5}")
```
